```python
from pyDatalog import pyDatalog

def load_fopc(file_path):    
    with open(file_path, "r", encoding="utf-8") as f:
        count = 0
        for line in f:
            if line.startswith("%"):
                continue
            print(line.strip())
            add_fact_from_line(line.strip())

def classify_number(s):
    try:
        i = int(s)
        return i
    except ValueError:
        try:
            f = float(s)
            return f
        except ValueError:
            return s

def add_fact_from_line(line):
    try:
        # Remove trailing period and whitespace
        line = line.strip().rstrip('.')
        # Extract predicate and arguments
        predicate, args_str = line.split('(', 1)
        args = args_str.rstrip(')')split(',')
        args = [classify_number(arg.strip()) for arg in args]
        
        # Build pyDatalog fact string
        fact = f"{predicate}({', '.join(repr(arg) for arg in args)})"
        # Evaluate the fact
        pyDatalog.assert_fact(predicate, *args)
        print(f"Added: {fact}")
    except Exception as e:
        print(f"Failed to add fact from line: {line}\nError: {e}")

# Example usage:
# add_fact_from_line("Ingredient(milk)")
# add_fact_from_line("NutrientValuePer100g(zucchini, fat, 0.205)")

```

```python
# Find recipe ingredients that are not declared with Ingredient(...)
from pyDatalog import pyDatalog
import pandas as pd

pyDatalog.create_terms('MissingIngredient, R, I')
# A recipe R has a missing ingredient I when Uses(R,I) holds but Ingredient(I) does not
MissingIngredient(R, I) <= Uses(R, I) & ~Ingredient(I)

print('MissingIngredient(R,I) results:')
print(MissingIngredient(R, I))

answers = pyDatalog.ask('MissingIngredient(R,I)')
if answers and getattr(answers, 'answers', None):
    df = pd.DataFrame(answers.answers, columns=['recipe','ingredient'])
    display(df)
else:
    print('No missing ingredients found (or no answers returned).')
```
